```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
import ast

#Data Loading
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Data Cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA = df[df['job_title_short'] == 'Data Analyst'].copy()
```

### Extract the month and creating its own column

```python
df_DA['job_posted_month_no'] = df_DA['job_posted_date'].dt.month
```

### Explode list into ceperated values

```python
df_DA_exploded = df_DA.explode('job_skills')
```

### Count each skill based on each month

```python
df_DA_pivot = df_DA_exploded.pivot_table(index='job_posted_month_no', columns='job_skills', aggfunc='size', fill_value=0)
df_DA_pivot
```

### Create Total row and sort by it the drop it

```python
df_DA_pivot.loc['Total'] = df_DA_pivot.sum()

df_DA_pivot = df_DA_pivot[df_DA_pivot.loc['Total'].sort_values(ascending=False).index]

df_DA_pivot = df_DA_pivot.drop('Total')

df_DA_pivot
```

### Convert month number to month name for ploting

```python
# convert month number to month name
df_DA_pivot['job_posted_month'] = pd.to_datetime(df_DA_pivot['job_posted_month_no'], format='%m').dt.strftime('%b')

df_DA_pivot.reset_index(inplace=True)
df_DA_pivot.set_index('job_posted_month', inplace=True)

df_DA_pivot.drop(['index','job_posted_month_no'], inplace=True)
```

```python
df_DA_pivot.iloc[:, :5].plot(kind='line', figsize=(12, 6), marker='o')
plt.title('Top 5 Trending Skills for Data Analysts Over Time')
```
