```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
import ast

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# Filter out rows with NaN values in the 'salary_year_avg' column
df_salary = df[pd.notna(df['salary_year_avg'])].copy()
```

### Apply calculation onto a column using function

```python
def projected_salary(salary):
    return salary * 1.03
```

```python
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)

df_salary.head(2)
```

```python
df_salary[['salary_year_avg','salary_year_inflated']].head()
```

### Turn String to List applying function

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)
    else:
        return skill_list
```

```python
df_salary['job_skills'] = df_salary['job_skills'].apply(clean_list)

df_salary.head(2)
```

### Apply Function on rows

```python
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else:
        return 1.03 * row['salary_year_avg']
```

```python
df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

df_salary.head(2)
```

```python
df_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']].head()
```

### If a salary match or dismash will return a result of High/Low on a new column

```python
def salary_category(x):
    if x >= 5000:
        return "High"
    else:
        return "Low"

df["salary_category"] = df["salary_year_avg"].map(salary_category)
```
