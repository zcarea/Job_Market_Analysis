### 🧠 **The mental map**

##### **When you receive a dirty dataset, think:**

1. **Inspect** → head(), info(), dtypes

2. **Missing** → isna(), fillna(), dropna()

3. **Duplicates** → duplicated(), drop_duplicates()

4. **Text** → .str.strip(), .str.lower(), .str.replace()

5. **Types** → to_numeric(), to_datetime()

6. **Weird value**s → unique(), value_counts(), replace()

7. **Bad rows** → filtering with df[...]

8. **Create/transform** → .map(), .dt...
