```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

### DataFrame based on country

```python
df_US = df[df['job_country'] == 'United States'].copy()
```

```python
# Add a new column for the month of the job posting
df_US['job_posted_month'] = df_US['job_posted_date'].dt.strftime('%B')
```

### Count posted roles based on months

```python
df_US_pivot = df_US.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
df_US_pivot
```

### Sort values by months numbers

```python
df_US_pivot.reset_index(inplace=True)

df_US_pivot['month_no'] = pd.to_datetime(df_US_pivot['job_posted_month'], format= '%B').dt.month
df_US_pivot.set_index('job_posted_month', inplace=True)
df_US_pivot.sort_values('month_no', inplace=True)
df_US_pivot
```

### Drop months columns after sorting values by it

```python
df_US_pivot.drop(columns='month_no', inplace=True)
df_US_pivot
```

### Finding Top 3 jobs to filter table with then ploting it

```python
top_jobs = df['job_title_short'].value_counts().head(3).index
top_jobs.tolist()

df_US_pivot[top_jobs].plot(kind='line')
```
