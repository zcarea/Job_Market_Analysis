```python
#EX1

class BaseSalary:
  def __init__( self, salary, bonus_rate=.1, symbol='$' ):
    self.salary = salary
    self.bonus_rate = bonus_rate
    self.symbol = symbol
    self.total_salary = self.salary * (1 + self.bonus_rate)
    self.bonus = self.total_salary - self.salary

# For clean Out-put (no need for it if you creat those methods)
  def __repr__(self):
    return f'{self.symbol}{self.total_salary:,.0f}'

  def show_salary(self):
    return f'{self.symbol}{self.total_salary:,.0f}'

  def show_bonus(self):
    return f'{self.symbol}{self.bonus:,.0f}'

salary = BaseSalary(100000)
salary.calculate_bonus()
```

```python
# Install datasets Library
# !pip install datasets

# Import Libraries
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
import pandas as pd

# Applying them onto columns
df['job_title'].count()

# Convert 'salary_year_avg' to numeric, coercing errors to handle non-numeric values
df['salary_year_avg'] = pd.to_numeric(df['salary_year_avg'], errors='coerce')

# To run median, the column must be numerical. NaNs will be ignored by default.
df['salary_year_avg'].median()

#---------- Investigation ------------------
""" Ex: a column has suspicious number probably low and see what row is it """
# 1st: Check its minimum id. Drop NaNs first to avoid errors with idxmin().
min_salary = df['salary_year_avg'].dropna().idxmin()

# 2nd: Reveal the full row data to see who has this min salary:
df.iloc[min_salary]
#-------------------------------------------

# Using value_counts can help to see how many time each value has been repeated:
df['job_title_short'].value_counts()

# What if we want to calculate min on each job:
# We group the column & select the column where the .min should be applied:
df.groupby('job_title_short')['salary_year_avg'].min()

# To group multiple columns and apply on them same agg:
df.groupby(['job_title_short','job_country'])['salary_year_avg'].min()

# To use multiple aggs in one code:
df.groupby('job_title_short')['salary_year_avg'].agg(['min','max'])
```
